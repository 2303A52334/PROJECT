**Bootstrappling**

```python
import numpy as np
import pandas as pd
def bootstrap(data, n_iterations=1000, statistic=np.mean, random_seed=42):
    np.random.seed(random_seed)
    bootstrap_stats = []
    for _ in range(n_iterations):
        sample = np.random.choice(data, size=len(data), replace=True)
        bootstrap_stats.append(statistic(sample))
    return np.array(bootstrap_stats)
adult_data = pd.DataFrame(data.data, columns=data.feature_names)
if 'education-num' in adult_data.columns:
    education_num_column = adult_data['education-num']
    n_iterations = 1000
    bootstrap_means = bootstrap(education_num_column, n_iterations=n_iterations, statistic=np.mean)
    confidence_level = 0.95
    lower_bound = np.percentile(bootstrap_means, (1 - confidence_level) / 2 * 100)
    upper_bound = np.percentile(bootstrap_means, (1 + confidence_level) / 2 * 100)
    print(f"Bootstrap mean estimate for 'education-num': {np.mean(bootstrap_means):.4f}")
    print(f"{confidence_level*100:.1f}% confidence interval: ({lower_bound:.4f}, {upper_bound:.4f})")
else:
    print("The dataset does not contain the 'education-num' column. Please check the dataset structure.")
```
